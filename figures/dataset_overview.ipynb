```python
import scanpy as sc
import pathlib
import yaml
```

```python
datasets = []
for config_path in pathlib.Path("../configs/signatures").iterdir():
    if config_path.stem in ["config", "debug"]:
        continue
    config = yaml.safe_load(open(config_path, "r"))
    for scenario, values in config["scenarios"].items():
        data_path = values["data_path"]
        datasets.append(data_path)
```

```python
n_cells = 0
n_samples = 0

for scenario in pathlib.Path().glob(pattern="../results/signatures/*"):
    if scenario.name in ["gbm_subsample", "sparse", "debug"]:
        continue
    for data_path in scenario.joinpath("data").joinpath("preprocessed").iterdir():
        if "all_" in data_path.name:
            continue
        if "scrna.h5ad" in data_path.name or "snrna.h5ad" in data_path.name:
            continue
        adata = sc.read_h5ad(data_path, backed="r")
        n_cells += adata.n_obs
        n_samples += adata.obs["sample"].nunique()


# /data/preprocessed
```
